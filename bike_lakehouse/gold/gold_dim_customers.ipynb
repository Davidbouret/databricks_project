#Init

```python
import pyspark.sql.functions as F
from pyspark.sql.window import Window
```

#Business transformations

```python
query = """
SELECT
    ci.customer_id,
    ci.customer_key,
    ci.first_name,
    ci.last_name,
    ci.gender,
    ci.marital_status
FROM silver.crm_customers ci
"""

df = spark.sql(query)
```

#Write into gold layer

```python
(df.write
    .mode("overwrite")
    .format("delta")
    .saveAsTable("gold.dim_customers")
 )
```
